```python
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import drive
drive.mount('/content/gdrive')

file_paths = [
    '/content/gdrive/My Drive/Colab Notebooks/LAB/data/csv_data/3_occ.csv',
    '/content/gdrive/My Drive/Colab Notebooks/LAB/data/csv_data/3_cu_1.csv',
    '/content/gdrive/My Drive/Colab Notebooks/LAB/data/csv_data/3_cu_2.csv',
    '/content/gdrive/My Drive/Colab Notebooks/LAB/data/csv_data/3_dev_11.csv',
    '/content/gdrive/My Drive/Colab Notebooks/LAB/data/csv_data/3_dev_12.csv',
    '/content/gdrive/My Drive/Colab Notebooks/LAB/data/csv_data/3_dev_21.csv',
    '/content/gdrive/My Drive/Colab Notebooks/LAB/data/csv_data/3_dev_22.csv'
]
for i, file_path in enumerate(file_paths, start=1):
  # 讀取 CSV 檔案
  df = pd.read_csv(file_path)

  # 篩選資料
  filtered_df = df.head(4000) #[df['Steps'] == 'DEV_TO_OCC']

  # 將 Timestamp 轉換為從 0 開始的時間
  filtered_df['Timestamp'] = filtered_df['Timestamp'] - filtered_df['Timestamp'].min()

  # 計算移動平均線（窗口大小為 5）
  filtered_df['CPU Usage (%) Moving Avg'] = filtered_df['CPU Usage (%)'].rolling(window=5).mean()

  # 繪製折線圖
  plt.figure(figsize=(3.25, 1.89))
  #plt.plot(filtered_df['Timestamp'], filtered_df['CPU Usage (%)'], label='CPU Usage (%)')
  plt.plot(filtered_df['Timestamp'], filtered_df['CPU Usage (%) Moving Avg'], label='CPU Usage (%) Moving Avg')
  plt.plot(filtered_df['Timestamp'], filtered_df['Memory Usage (%)'], label='Memory Usage (%)')
  plt.xlabel('Timestamp')
  plt.ylabel('Percentage')
  plt.ylim(0, 125)
  #plt.title(f'CPU and Memory Usage Over Time - {file_path.split("/")[-1].split(".")[0]}')

  # 計算平均值並顯示在圖表上
  cpu_avg = filtered_df['CPU Usage (%)'].mean()
  memory_avg = filtered_df['Memory Usage (%)'].mean()
  print(f'CPU and Memory Usage Over Time - {file_path.split("/")[-1].split(".")[0]}')
  print(f'CPU Avg: {cpu_avg:.2f}')
  print(f'Memory Avg: {memory_avg:.2f}')
  #plt.text(1.1, 0.9, f'CPU Avg: {cpu_avg:.2f}%', transform=plt.gca().transAxes)
  #plt.text(1.1, 0.85, f'Memory Avg: {memory_avg:.2f}%', transform=plt.gca().transAxes)

  plt.legend()
  plt.show()
```

```python
import re
import pandas as pd
from google.colab import drive
drive.mount('/content/gdrive')

# 讀取CSV文件
df = pd.read_csv('/content/gdrive/My Drive/Colab Notebooks/LAB/data/csv_datat/3_occ_t.csv')

# 選擇 Steps 為 DEV_TO_OCC 的行
df = df[df['Steps'] == 'DEV_TO_OCC']

# 初始化一個列表來存儲差值
diff_list = []

# 編譯正則表達式模式（只編譯一次以提高效率）
pattern = re.compile(r"'11001'\s*:\s*'(,\d+|\d+,\d+)'")

# 遍歷每一行
for index, row in df.iterrows():
    try:
        # 使用正則表達式查找 '11001' 的值
        match = pattern.search(row['DataTimestamp'])
        if match:
            value = match.group(1)
            # 去掉逗號前面的部分
            parts = value.split(',')
            extracted_value = parts[-1]  # 取最後一部分，無論是否有逗號

            # 檢查提取的值是否為數字
            if extracted_value.isdigit():
              # 計算差值
              diff = row['Timestamp'] - float(extracted_value)
              diff_list.append(diff)
    except (AttributeError, IndexError, ValueError):
        # 如果出現任何錯誤（屬性錯誤、索引錯誤或值轉換錯誤），則跳過此行
        pass

print(len(diff_list))
# 將差值轉換為 Series 並進行統計分析
diff_series = pd.Series(diff_list[300:400])
result = diff_series.describe()

print(result)
```
